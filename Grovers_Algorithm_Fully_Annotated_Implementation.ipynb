# Grover's Algorithm – Fully Annotated Implementation

This notebook walks through Grover's Algorithm step by step with explanations for every major section.

## Step 1 – Imports

```
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator
from qiskit.transpiler.preset_passmanagers import generate_preset_pass_manager
```

These imports provide the quantum circuit classes, simulator, and transpiler used throughout the program.

## Step 2 – Create Registers

```
qr = QuantumRegister(3, 'q')
cr = ClassicalRegister(3, 'c')
qc = QuantumCircuit(qr, cr)
```

Initially every qubit is in the state **|0⟩**, so the complete system starts in **|000⟩**.

## Step 3 – Create Superposition

```
qc.h(qr)
```

Applying a Hadamard gate to each qubit creates an equal superposition of all eight basis states.

## Step 4 – Barrier

```
qc.barrier()
```

A barrier has no computational effect; it simply improves circuit readability.

## Step 5 – Oracle

```
# Example oracle that marks |101>
qc.x(1)

qc.h(2)
qc.ccx(0,1,2)
qc.h(2)

qc.x(1)
```

The oracle marks the desired solution by applying a phase flip. It does **not** measure the qubits.

## Step 6 – Diffusion Operator

```
qc.h(qr)
qc.x(qr)

qc.h(2)
qc.ccx(0,1,2)
qc.h(2)

qc.x(qr)
qc.h(qr)
```

The diffusion operator amplifies the probability of measuring the marked state.

## Step 7 – Measurement

```
qc.measure(qr, cr)
```

Measurement collapses the quantum state into classical bits. It should occur after the oracle and diffusion operator.

## Step 8 – Execute

```
backend = AerSimulator()

pm = generate_preset_pass_manager(
    backend=backend,
    optimization_level=1
)

compiled = pm.run(qc)

job = backend.run(compiled, shots=1024)

result = job.result()
counts = result.get_counts()

print(counts)
```

# Summary

Workflow:

1. Create registers
2. Apply Hadamard gates
3. Build the Oracle
4. Apply the Diffusion Operator
5. Measure
6. Execute
7. Analyze counts

Remember:
- Oracle marks the answer.
- Diffusion amplifies the answer.
- Measurement reveals the answer.
